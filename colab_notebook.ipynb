## 🔄 Alternative: Direct Python Usage

You can also use the face swapping directly in code:

```python
# Example: Direct face swapping in code
from face2face.core.face2face import Face2Face
import cv2
from google.colab import files
import matplotlib.pyplot as plt

# Initialize model
print("Loading Face2Face model...")
f2f = Face2Face(device_id=0)
print("✅ Model loaded!")

# Upload images
print("\nUpload your movie poster:")
poster_upload = files.upload()
poster_path = list(poster_upload.keys())[0]

print("\nUpload your couples photo:")
couple_upload = files.upload()
couple_path = list(couple_upload.keys())[0]

# Load images
poster_img = cv2.imread(poster_path)
couple_img = cv2.imread(couple_path)

print("\n🔄 Swapping faces...")
# Perform face swap
result = f2f.swap(
    media=(couple_img, poster_img),
    enhance_face_model='gpen_bfr_512'
)

# Get result image
if hasattr(result, 'get_image'):
    swapped_img = result.get_image()
else:
    swapped_img = result

# Display result
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(poster_img, cv2.COLOR_BGR2RGB))
plt.title('Original Poster')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(couple_img, cv2.COLOR_BGR2RGB))
plt.title('Couples Photo')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(swapped_img, cv2.COLOR_BGR2RGB))
plt.title('Swapped Result')
plt.axis('off')

plt.tight_layout()
plt.show()

# Save result
cv2.imwrite('swapped_poster.jpg', swapped_img)
print("\n✅ Done! Downloading result...")
files.download('swapped_poster.jpg')
```

## 🆘 Troubleshooting

### "Killed" or OOM Error
- Runtime → Factory reset runtime
- Make sure GPU is enabled
- Try using smaller images

### Models not downloading
- Check internet connection
- Wait a few minutes (models are ~500MB)
- Check for error messages in output

### Gradio link not working
- Make sure the cell is still running
- Check if session is still active
- Try rerunning the launch cell
